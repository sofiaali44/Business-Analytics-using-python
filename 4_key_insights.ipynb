##### KEY INSIGHTS: WHY ARE CUSTOMERS LEAVING?

• Customers on month-to-month contracts have the highest churn risk,
  suggesting low commitment customers are easier to lose.

• Fiber optic customers show higher churn compared to other internet services,
  indicating possible service quality or pricing concerns.

• Customers without technical support have higher churn,
  highlighting the importance of customer assistance and issue resolution.

• Churned customers generally have shorter tenure,
  showing that early customer experience is critical for retention.

##### RECOMMENDATION:
Focus retention efforts on new customers, month-to-month users,
and customers without technical support through improved onboarding,
support services, and targeted retention offers.
